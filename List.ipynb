#5.Copy()

```python
original_list = [10,20,30]
print(original_list)
```

```python
# This will used to create duplicate copy of the existing list
#Deep copy
#let me create a deep copy from the original list
copy_list = original_list
```

```python
#Cross check 
print("Original data ",original_list)
print("The copy data ",copy_list)
```

```python
#let me add one data in copy?
copy_list.append("Sayli")
```

```python
#Cross check
print("Original data ",original_list)
print("The copy data ",copy_list)
```

**Note**:Avoid Deep Copy

```python
#Shallow Copy
original_list
```

```python
#let me create a shallow copy using the original list
new_copy_list = original_list.copy()
```

```python
#Cross check
print("Original data ",original_list)
print("The copy data ",new_copy_list)
```

```python
#Now let me add one element to the copy
new_copy_list.append("Khedkar")
```

```python
#Cross check
print("Original data ",original_list)
print("The copy data ",new_copy_list)
```
