### Q13. [Replacing Values] Replace all missing values in `sales_with_missing` with the mean of the available values. Store the cleaned result in a new array.

```python
sales_with_missing = np.array([
    12000, 18000, np.nan, 15000, 30000, np.nan, 40000
])
```

```python
# Calculate mean ignoring missing values
mean_sales = np.nanmean(sales_with_missing)
```

```python
# Create a new array and replace missing values
cleaned_sales = np.where(
    np.isnan(sales_with_missing),
    mean_sales,
    sales_with_missing
)
```

```python
mean_sales
```

```python
cleaned_sales
```

### Q19. [Correlation] Calculate the correlation coefficient between `advertising_spend` and `revenue`. Based only on the coefficient, state whether the relationship is positive or negative and whether it appears weak or strong.

```python
advertising_spend = np.array([10000, 20000, 30000, 40000, 50000])
revenue = np.array([50000, 70000, 90000, 110000, 130000])
```

```python
# Calculate correlation coefficient
correlation = np.corrcoef(advertising_spend, revenue)[0, 1]
```

```python
print("Correlation coefficient:", correlation)
```

```python
if correlation > 0:
    print("Relationship: Positive")
else:
    print("Relationship: Negative")
```

```python
if abs(correlation) >= 0.7:
    print("Strength: Strong")
else:
    print("Strength: Weak")
```

### Q20. [Interview Challenge] Using the `sales` array, calculate the percentage contribution of every transaction to total sales, identify the transaction with the highest contribution, and return only transactions whose contribution is above the average contribution

```python
sales = np.array([12000, 18000, 25000, 15000, 30000, 22000, 40000])
```

```python
# 1. Calculate total sales
total_sales = np.sum(sales)
total_sales
```

```python
# 2. Calculate percentage contribution of each transaction
contribution = (sales / total_sales) * 100
contribution
```

```python
# 3. Find transaction with highest contribution
highest_index = np.argmax(contribution)
highest_transaction = sales[highest_index]
highest_index
```

```python
highest_transaction
```

```python
# 4. Calculate average contribution
average_contribution = np.mean(contribution)
average_contribution
```

```python
# 5. Select transactions above average contribution
above_average = sales[contribution > average_contribution]
above_average
```
