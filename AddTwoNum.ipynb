```python
class ListNode:
    def __init__(self,val=0,next=None):
        self.val = val
        self.next = next
        
class Solution:
    def addTwoNumbers(self,l1,l2):
        dummy = ListNode(0)
        current = dummy
        carry = 0

        while l1 or l2 or carry:
            if l1:
                x=l1.val
                l1 = l1.next
            else:
                x = 0

            if l2:
                y = l2.val
                l2 = l2.next
            else:
                y = 0

            total = x+y+carry
            carry = total//10

            current.next = ListNode(total%10)
            current = current.next
        return dummy.next

l1 = ListNode(2)
l1.next = ListNode(4)
l1.next.next = ListNode(3)

l2 = ListNode(5)
l2.next = ListNode(6)
l2.next.next = ListNode(4)

obj = Solution()
result = obj.addTwoNumbers(l1,l2)

while result:
    print(result.val,end=" ")
    result = result.next
```
