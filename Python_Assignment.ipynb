## Q39. [Mixed Scenario] Create a report showing each employee name, department, salary, and salary category. Salary category: salary >= 70000 is 'High', salary >= 50000 is 'Medium', otherwise 'Low'.

```python
# Write your answer here
for name, department in zip(employee_names, departments):
    salary = employee_salary.get(name, 0)
    if salary >= 70000: 
        category = "High" 
    elif salary >= 50000: 
        category = "Medium" 
    else: category = "Low" 
    print(name, department, salary, category)
```

## Q40. [Mixed Scenario] Business case: Using `sales_amounts`, identify duplicate sales values, unique sales values, total sales, and top 3 sales values. Store the output in a dictionary called `sales_report`.

```python
# Write your answer here
# Find duplicate values 
duplicates = []
for sale in sales_amounts: 
    if sales_amounts.count(sale) > 1 and sale not in duplicates:
        duplicates.append(sale)
# Unique sales 
unique_sales = list(set(sales_amounts)) 
# Total sales
total_sales = sum(sales_amounts)
# Top 3 sales
top_3_sales = sorted(sales_amounts, reverse=True)[:3]
# Create report 
sales_report = { "duplicates": duplicates, "unique_sales": unique_sales, "total_sales": total_sales, "top_3_sales": top_3_sales }
print(sales_report)
```
